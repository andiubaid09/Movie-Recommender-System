**Recommendation Ruled-Based Popularity**

```python
def popularity_recommender(data, top_n=10):
  recommendations = (
      data[['title','popularity','genres','vote_average','overview']].sort_values(by='popularity',ascending=False).head(10)
  )
  return recommendations.reset_index(drop=True)
```

```python
def display_recommendations(recommedations):
  for index, row in recommedations.iterrows():
    print(f"{index + 1 }. {row['title']}")
    print(f"   Rating     : {row['vote_average']}")
    print(f"   Popularity : {row['popularity']}")
    print(f"   Genre      : {row['genres']}")
    print(f"   Overview   : {row['overview']}")
    print("-"*200)
```

```python
recommendation = popularity_recommender(recommendation_df, top_n=10)
display_recommendations(recommendation)
```
